# 🚀 Dynamic Programming (DP)

## 1. What is Dynamic Programming?

**Dynamic Programming (DP)** is a technique used to solve problems by breaking them into smaller subproblems and storing their answers so we don't have to solve the same subproblems again.

💡 **Key Idea:** Store the answers of already solved subproblems and reuse them whenever needed.

### 📌 Two Approaches to Dynamic Programming

1. **Memoization (Top-Down Approach)**

2. **Tabulation (Bottom-Up Approach)**

---

## 2. Memoization (Top-Down Approach)

**Memoization** means storing the result of a function so that we don't calculate it again.

* Uses **recursion** to solve the problem.

* Stores previously calculated results in a **dictionary or array**.

* Checks whether the result is already stored before calculating it again.

* Starts with the original problem and breaks it into smaller subproblems.


## 3. Tabulation (Bottom-Up Approach)

**Tabulation** means solving smaller problems first and storing their answers in a table (usually an array).

* Uses **loops** instead of recursion.

* Stores results in an array or table.

* Starts with the smallest subproblems and builds up to the original problem.

* Generally avoids recursive function calls.


### 💡 Example: Fibonacci using Tabulation

```
def fib(n):
    dp = [0] * (n + 1)

    dp[0] = 0

    if n >= 1:
        dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n]

print(fib(5))  # Output: 5
```

**How it works:**

* First, initialize the base cases: `dp[0] = 0` and `dp[1] = 1`.

* Use a loop to calculate the next Fibonacci numbers.

* Store each result in the `dp` array.

* Finally, return `dp[n]`.

---

## 4. Difference Between Memoization and Tabulation

| Feature        | Memoization           | Tabulation                             |
| -------------- | --------------------- | -------------------------------------- |
| Approach       | Top-Down              | Bottom-Up                              |
| Technique      | Recursion             | Iteration (loops)                      |
| Starting point | Original problem      | Smallest subproblem                    |
| Storage        | Dictionary or array   | Array or table                         |
| Calculation    | On demand             | Usually calculates all required states |
| Recursion      | Yes                   | No                                     |
| Stack usage    | Uses recursion stack  | No recursion stack                     |
| Implementation | Often easier to write | Often more efficient in practice       |

---

## ⭐ Quick Revision

🔹 **Memoization:** Recursion + Caching (Top-Down)

🔹 **Tabulation:** Loops + Table (Bottom-Up)

🔹 **Main purpose of DP:** Avoid repeated calculations by storing and reusing results.

🔹 **Both approaches:** Solve overlapping subproblems and can reduce the time complexity of a problem.

📌 **Remember:** Memoization starts with the big problem and moves toward smaller problems, while tabulation starts with smaller problems and builds toward the final answer.


### 0 - 1 Knapsack Problem

#### 🔗 Practice Link  
👉  [0/1 Knapsack — GeeksforGeeks](https://www.geeksforgeeks.org/problems/0-1-knapsack-problem0945/1)

````python
class Solution:
    def knapsack(self, W: int, val: list[int], wt: list[int]) -> int:
        n = len(val)

        dp = [[0] * (W + 1) for _ in range(n + 1)]

        for i in range(1, n + 1):
            for capacity in range(1, W + 1):

                if wt[i - 1] <= capacity:
                    pick = val[i - 1] + dp[i - 1][capacity - wt[i - 1]]
                    notPick = dp[i - 1][capacity]

                    dp[i][capacity] = max(pick, notPick)

                else:
                    dp[i][capacity] = dp[i - 1][capacity]

        return dp[n][W]
    
## Time complexity O(NW) N = number of items w = knapsack capacity
## space complexity O(NW) 
````

### Coin Change

#### 🔗 Practice Link  
👉  [Coin Change — LeetCode](https://leetcode.com/problems/coin-change/description/)

````python
class Solution:
    def coinChange(self, coins, amount):
        dp = [float('inf')] * (amount + 1)

        dp[0] = 0

        for i in range(1, amount + 1):
            for coin in coins:
                if coin <= i:
                    dp[i] = min(dp[i], 1 + dp[i - coin])

        if dp[amount] == float('inf'):
            return -1

        return dp[amount]
    
## Time Complexity O(A * N) a = amount n = number of coins
## Auxilary space  O(A)
````

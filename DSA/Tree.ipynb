# 🌳 Binary Search Tree (BST) — Node Structure

In a **Binary Search Tree (BST)**, each node has **3 fields**:

1. 🔑 **Key / Data**
   → Stores the value of the node.

2. ⬅️ **Left Reference**
   → Stores the reference to the left child.

3. ➡️ **Right Reference**
   → Stores the reference to the right child.

### 🌱 Example

```text
       50
      /  \
    30    70
````

### 📌 For node 50:

```text
Key   = 50
Left  = Reference to 30
Right = Reference to 70
```

### 📌 For node 30:

```text
Key   = 30
Left  = None
Right = None
```

If a node does **not have a left or right child**, that reference is stored as **None**.

### 🧠 Easy Way to Remember

```text
Left Child Reference  +  Key/Data  +  Right Child Reference
```

⭐ **BST Node = Left Reference + Data + Right Reference**

### Inorder traversal in python

`````python
class Node :
    
    def __init__(self,k) :
        
        self.key = k
        self.left = None
        self.right = None
        
        
def inorder(root) :
    
    if root != None :
        inorder(root.left)
        print(root.key)
        inorder(root.right)
        
        
        
root = Node(10)

root.left = Node(20)
root.right = Node(30)
root.right.left = Node(40)
root.right.right = Node(50)

inorder(root)

# Time Complexity theta(n)
# space compl;exity theta(h)
        
`````

`````python
def preorder(root) :
    
    if root != None :
        print(root.key)
        preorder(root.left)
        preorder(root.right)
        
preorder(root)        
`````

`````python
def postorder(root) :
    
    if root != None :
        postorder(root.left)
        postorder(root.right)
        print(root.key)
        
        
postorder(root)
`````

`````python
def height(root) :
    
    if root == None :
        return 0
    
    else :
        
        lh = height(root.left)
        rh = height(root.right)
        return max(lh,rh) + 1
    
    
height(root)

# Time Complexity theta(n)
# space compl;exity theta(h)
`````

`````python
def printkdist(root,k) :
    
    if root == None :
        return
    
    if k == 0 :
        print(root.key, end = " ")
        
    else :
        printkdist(root.left,k-1)
        printkdist(root.right,k-1)
        
        

printkdist(root,2)        
`````

`````python
from collections import deque

def printlevelorder(root) :
    
    if root is None :
        return
    
    q = deque()
    q.append(root) 
    
    while len(q) > 0 :
        
        node = q.popleft()
        
        print(node.key, end = " ")
        
        if node.left is not None :
            q.append(node.left)
            
        if node.right is not None :
            q.append(node.right)
            
            
printlevelorder(root)
`````

#### Size of Binary tree in python

`````python
def treesize(root) :
    
    if root == None :
        return 0
    
    else :
        
        ls = treesize(root.left)
        rs = treesize(root.right)
        return ls+rs+1
    
treesize(root)
`````

#### Maximum in Binary tree

`````python
import math

def getmax(root) :
    
    if root == None :
        return -math.inf
    
    else :
        
        lm = getmax(root.left)
        rm = getmax(root.right)
        return max(root.key,lm,rm)
    
getmax(root)
`````

#### Search in BST

`````python
def searchbst(root, key) :
    
    if root is None :
        return False
    
    elif root.key == key :
        return True
    
    elif root.key > key :
        return searchbst(root.left, key)
    
    else :
        return searchbst(root.right, key)
    
    
searchbst(root,30)        
`````

`````python
def searchbst(root, key) :
    
    while root != None :
        
        if root.key == key :
            return True
        
        elif root.key > key :
            root = root.left
        
        else :
            root = root.right
            
    return False

searchbst(root,30)   
`````

#### Insert in BST

`````python
class Node :
    
    def __init__(self,key) :
        
        self.key = key
        self.left = None
        self.right = None
        
root = Node(10)
root.left = Node(5)
root.left.right = Node(7)
root.right = Node(15)
root.right.left = Node(12)
root.right.right = Node(18)
        
`````

`````python
def preorder(root) :
    
    if root != None :
        
        print(root.key)
        preorder(root.left)
        preorder(root.right)
    
    
preorder(root)
`````

`````python
def insert(root, key) :
    
    if root == None :
        return Node(key)
    
    if root.key == key :
        return root
    
    elif root.key > key :
        root.left = insert(root.left,key)
    
    else :
        root.right = insert(root.right,key)
        
    return root
    
    
insert(root,3)
preorder(root)
`````

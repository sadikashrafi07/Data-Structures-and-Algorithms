# Sieve of Eratosthenes

The Sieve of Eratosthenes is an efficient algorithm used to find all prime numbers up to a given number `n`.

It works by:

- Assuming all numbers are prime initially
- Repeatedly marking multiples of prime numbers as non-prime
- The remaining unmarked numbers are prime numbers

### Time Complexity

```python
O(n log log n)
```

### Space Complexity

```python
O(n)
```

### Best Use

- Finding all prime numbers in a range
- Competitive programming
- Large input values

````python
def sieve(n):

    prime = [True] * (n + 1)

    prime[0] = prime[1] = False

    for i in range(2, int(n**0.5) + 1):

        if prime[i]:

            for j in range(i * i, n + 1, i):
                prime[j] = False

    for i in range(2, n + 1):

        if prime[i]:
            print(i, end=" ")


n = int(input("Enter a number: "))
sieve(n)

#Time Complexity O(n log log n)
# Space Complexity O(n)
````

### Iterative power(Binary Exponentiation)

````python
def power(x,n):
    res = 1
    while n > 0:
        if n % 2 != 0:
            res = res * x
        x = x * x
        n = n // 2
    return res

if __name__ == '__main__':
    x = 4
    n = 5
    print(power(x, n))
````

Initial Values
- x = 4, n = 5, res = 1

First Iteration  
- n is odd (5 % 2 != 0), so res = res * x => res = 1 * 4 => res = 4
- x is squared: x = x * x => x = 4 * 4 => x = 16
- n is halved: n = n // 2 => n = 5 // 2 => n = 2
    
Second Iteration  
- n is even (2 % 2 == 0), so res remains unchanged: res = 4
- x is squared: x = x * x => x = 16 * 16 => x = 256
- n is halved: n = n // 2 => n = 2 // 2 => n = 1
    
Third Iteration  
- n is odd (1 % 2 != 0), so res = res * x => res = 4 * 256 => res = 1024
- x is squared: x = x * x => x = 256 * 256 => x = 65536
- n is halved: n = n // 2 => n = 1 // 2 => n = 0
    
Loop Ends  
n is now 0, so the loop exits and the function returns res, which is 1024.

````python
def compress_string(s):
    if not s:  # Handle the case of an empty string
        return ""

    compressed = []  # To store the compressed version of the string
    count = 1  # Start counting from the first character
    current_char = s[0]  # Initialize the first character

    # Iterate through the string starting from the second character
    for i in range(1, len(s)):
        if s[i] == current_char:
            count += 1  # Increment count if the same character continues
        else:
            # Append the current character and its count (if greater than 1) to the result
            compressed.append(current_char)
            if count > 1:
                compressed.append(str(count))
            # Update current_char and reset count for the new character
            current_char = s[i]
            count = 1

    # Append the last character and its count (if greater than 1) to the result
    compressed.append(current_char)
    if count > 1:
        compressed.append(str(count))

    return ''.join(compressed)  # Join the list into a single string

# Example usage
if __name__ == '__main__':
    input_str1 = "aabbbbeeeeffggg"
    input_str2 = "abbccccc"
    
    print(compress_string(input_str1))  # Output: a2b4e4f2g3
    print(compress_string(input_str2))  # Output: ab2c5
````

#### 🧠 Problem:
- In the given code, the line `compressed.append(count)` adds an integer to the list.
- But later, when we do ''.join(compressed), it expects all elements to be strings.
- Hence, Python will raise a TypeError.

''.join(compressed)  # This will cause an error because of integer 2


Consider the input string "aabbbcc":   

The loop starts at index 1 (the second character):   

##### First iteration (i = 1, s[i] = 'a'):
- if s[i] == current_char → True (because s[i] = 'a' and current_char = 'a'), so count = 2.  

##### Second iteration (i = 2, s[i] = 'b'):
- if s[i] == current_char → False (because s[i] = 'b' and current_char = 'a'), so the else block runs :  
- compressed.append('a') and compressed.append('2').  
- Now, compressed = ['a', '2'].  
- current_char = 'b' and count = 1.  
The loop continues with current_char = 'b':  

##### Third iteration (i = 3, s[i] = 'b'):
- if s[i] == current_char → True, so count = 2.

##### Fourth iteration (i = 4, s[i] = 'b'):
- if s[i] == current_char → True, so count = 3.

##### Fifth iteration (i = 5, s[i] = 'c'):

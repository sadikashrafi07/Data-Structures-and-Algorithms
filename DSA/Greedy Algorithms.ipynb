# Greedy Algorithm in DSA

## 1. What is a Greedy Algorithm?

A **Greedy Algorithm** is an algorithm that makes the **best possible choice at each step** to solve a problem.

It chooses the option that looks best at the current moment without worrying too much about future steps.

**Main idea:** Make the best choice now and try to get the best overall solution.

**Example:**

Suppose you have ₹18 and coins of ₹1, ₹2, ₹5, and ₹10.

To make ₹18 using the minimum number of coins:

1. Choose ₹10. Remaining amount = ₹8.

2. Choose ₹5. Remaining amount = ₹3.

3. Choose ₹2. Remaining amount = ₹1.

4. Choose ₹1. Remaining amount = ₹0.

**Answer:** ₹10 + ₹5 + ₹2 + ₹1 = ₹18 (4 coins).

At every step, we choose the largest possible coin. This is a greedy approach.

---

## 2. How Does a Greedy Algorithm Work?

A greedy algorithm generally follows these steps:

1. Start with an empty solution.

2. Find all the available choices.

3. Choose the best option at the current step.

4. Add the selected option to the solution.

5. Repeat until the problem is solved.

**In simple words:** Choose the best option available now and continue until you get the final answer.

---

## 3. Important Properties of Greedy Algorithms

### 1. Greedy Choice Property

A problem has the greedy-choice property if we can make the best choice at each step and still obtain an optimal overall solution.

**In simple words:** Choosing the best option now should not prevent us from getting the best final answer.

### 2. Optimal Substructure

A problem has optimal substructure if its optimal solution can be constructed using optimal solutions to its smaller subproblems.

**In simple words:** We can build the best overall solution by solving smaller parts optimally.

These two properties help us identify whether a greedy approach is suitable for a problem.

---

## 4. Example: Coin Change Using Greedy

**Problem:** Find the minimum number of coins needed to make an amount of 18.

Coins = `[1, 2, 5, 10]`

### Python Code

```
coins = [10, 5, 2, 1]
amount = 18
count = 0

for coin in coins:
    while amount >= coin:
        amount -= coin
        count += 1

print(count)
```

**Output:**

```
4
```

### Dry Run

| Coin | Remaining Amount | Number of Coins |
| ---- | ---------------- | --------------- |
| 10   | 8                | 1               |
| 5    | 3                | 2               |
| 2    | 1                | 3               |
| 1    | 0                | 4               |

**Final Answer:** 4 coins (10 + 5 + 2 + 1).

**Explanation:**

* We start with the largest coin.

* If the coin is less than or equal to the remaining amount, we select it.

* We subtract its value from the amount.

* We repeat until the remaining amount becomes zero.

**Note:** This greedy approach works for this set of coins, but it does not work for every coin system.

---

## 5. When Does Greedy Fail?

Greedy does not always give the optimal solution.

**Example:**

Coins = `[1, 3, 4]`

Amount = `6`

Using the greedy approach:

1. Choose 4. Remaining amount = 2.

2. Choose 1. Remaining amount = 1.

3. Choose 1. Remaining amount = 0.

Greedy answer = 3 coins (4 + 1 + 1).

But the optimal answer is:

3 + 3 = 6

Optimal answer = 2 coins.

Therefore, greedy does not always give the minimum number of coins.

**Important:** Always check whether the greedy choice leads to an optimal solution.

## 6. Greedy Algorithm vs Dynamic Programming

| Greedy Algorithm                              | Dynamic Programming                                    |
| --------------------------------------------- | ------------------------------------------------------ |
| Makes the best choice at each step.           | Solves smaller subproblems and combines their results. |
| Usually does not reconsider previous choices. | Stores and reuses previously calculated results.       |
| Often simpler and faster.                     | May require more time and memory.                      |
| Does not always give the optimal answer.      | Gives an optimal answer when correctly formulated.     |
| Example: Activity Selection.                  | Example: 0/1 Knapsack.                                 |

**Main difference:**

* **Greedy:** Makes the best choice at the current step.

* **Dynamic Programming:** Considers smaller subproblems and combines their results to find the optimal solution.

## 8. Advantages of Greedy Algorithms

1. Easy to understand and implement for suitable problems.

2. Often require less memory.

3. Can be faster than dynamic programming.

4. Useful for optimization problems.

5. Can produce an optimal solution when the greedy strategy is valid.

## 9. Disadvantages of Greedy Algorithms

1. They do not always produce an optimal solution.

2. A locally optimal choice may lead to a poor overall solution.

3. Previous choices are generally not reconsidered.

4. Proving that a greedy approach is correct can be difficult.

---

## 10. When Should We Use Greedy Algorithms?

Consider a greedy approach when:

1. The problem asks you to maximize or minimize something.

2. You can identify the best choice at each step.

3. Making a choice does not prevent you from reaching the optimal solution.

4. The problem has the greedy-choice property and optimal substructure.

5. You can prove that the greedy approach produces an optimal answer.

If greedy does not work, consider dynamic programming or another suitable algorithm.

---

## 11. Important Points for Interviews

* Greedy algorithms make the best local choice at every step.

* A locally optimal choice is not necessarily globally optimal.

* Greedy algorithms are commonly used in optimization problems.

* Always check whether the greedy strategy gives the optimal solution.

* Sorting is often used to determine the order of greedy choices.

* Greedy algorithms can have different time complexities depending on the problem.

* Common greedy problems include Activity Selection, Fractional Knapsack, Job Sequencing, and Minimum Spanning Trees.

### How to Identify a Greedy Problem

Ask yourself these questions:

1. What is the best choice I can make at the current step?

2. Can I make this choice without reconsidering it later?

3. Will this choice help me achieve the optimal final answer?

4. Can I prove that my greedy strategy is correct?

If you can justify these choices, greedy may be an appropriate approach.

### Minimum Coins of 1, 2, 5 and 10

#### 🔗 Practice Link  
👉  [Minimum Coins of 1, 2, 5 and 10— GeeksforGeeks](https://www.geeksforgeeks.org/problems/-minimum-number-of-coins4426/1)

````python
class Solution:
    def findMin(self, n: int) -> int:
       # code here 
       coins = [1,2,5,10]
       coins.sort(reverse = True)
       
       res = 0
       
       for x in coins :
           
           if x <= n :
               
               c = n // x
               res += c
               n -= c*x
               
           if n == 0 :
               break
           
       return res
               
````

### Activity Selection

#### 🔗 Practice Link  
👉  [Activity Selection — GeeksforGeeks](https://www.geeksforgeeks.org/problems/activity-selection-1587115620/1)

````python
class Solution:
    def activitySelection(self, start: list[int], finish: list[int]) -> int:

        arr = []

        for i in range(len(start)):
            arr.append((start[i], finish[i]))

        arr.sort(key=lambda x: x[1])

        prev = 0
        res = 1

        for curr in range(1, len(arr)):

            if arr[curr][0] > arr[prev][1]:
                res += 1
                prev = curr

        return res
````

### Fractional Knapsack Problem

#### 🔗 Practice Link  
👉  [Fractional Knapsack — GeeksforGeeks](https://www.geeksforgeeks.org/problems/fractional-knapsack-1587115620/1)

````python
class Solution:
    def fractionalKnapsack(self, val, wt, capacity):
        
        n = len(val)
        costweight = []
        
        # Create (value, weight, value/weight)
        for i in range(n):
            c, W = val[i], wt[i]
            costweight.append((c, W, (c * 1.0) / W))
        
        # Sort by value/weight ratio in descending order
        costweight = sorted(costweight, key=lambda x: x[2], reverse=True)
        
        res = 0.0
        
        # Pick items
        for curr in costweight:
            
            # If complete item can fit
            if curr[1] <= capacity:
                res += curr[0]
                capacity -= curr[1]
            
            # If only a fraction can fit
            else:
                res += curr[0] * (capacity / curr[1])
                break
        
        return round(res, 6)
````
